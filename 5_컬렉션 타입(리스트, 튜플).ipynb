# **1. 컬렉션 타입**
여러 개의 데이터 항목을 하나의 단위로 관리할 수 있게 해주는 데이터 구조를 의미합니다. 이것은 여러 개의 데이터를 하나의 변수에 저장하고, 관리할 수 있게 해줍니다. 파이썬에서는 리스트, 튜플, 세트, 딕셔너리 등이 기본적인 컬렉션 타입에 속합니다.

![컬렉션 타입](./images/컬렉션%20타입.png)

# **2. 리스트**
리스트(List)는 여러 개의 데이터를 하나의 변수에 순서대로 저장할 수 있는 Python의 대표적인 컬렉션 자료형으로, 대괄호([])를 사용해 생성합니다. 각 요소는 인덱스를 통해 접근할 수 있으며, 서로 다른 자료형의 값을 함께 저장할 수 있고, 요소의 추가·수정·삭제가 가능한 mutable(변경 가능한) 구조를 가집니다. 이러한 특징으로 인해 리스트는 데이터를 묶어서 관리하거나 반복 처리할 때 매우 유용하게 사용됩니다.

![리스트](./images/리스트.png)

````python
li1 = [1, 3, 5, 7, 9]
print(li1)
print(type(li1))

print('--------------------------------------')

li2 = list([1, 3, 5, 7, 9])
print(li2)
print(type(li2))

print('--------------------------------------')

li3 = ['김사과', '반하나', '오렌지', '이메론']
print(li3)

print('--------------------------------------')

li4 = [1, 50.5, '김사과', True] # 각자 타입이 다른 데이터를 저장
print(li4)
print(type(li4))

print(li4[0])
print(type(li4[0]))
print(li4[1])
print(type(li4[1]))
print(li4[2])
print(type(li4[2]))
print(li4[3])
print(type(li4[3]))
# print(li4[4]) # IndexError: list index out of range
````

### 리스트 인덱싱
- 리스트의 각 항목은 위치(인덱스)를 가지고 있으며, 인덱스를 사용하여 접근할 수 있습니다. 
- 인덱스는 0부터 시작합니다.
- 인덱싱은 리스트에서 '하나의 요소'를 꺼내는 것입니다.
- 결과는 → 꺼낸 하나의 값(값 자체)입니다.
- 대괄호로 감싸지지 않습니다.

````python
li1 = [1, 3, 5, 7, 9]
print(li1)
print(li1[0])
print(li1[-1])
print(li1[0] + li1[-1])

print('--------------------------------------')

li2 = [1, 2, '파이썬', ['김사과', '오렌지']]
print(li2)
print(type(li2))
print(li2[1])
print(type(li2[1]))
print(li2[3])
print(type(li2[3]))
print(li2[3][1])
print(type(li2[3][1]))

print('--------------------------------------')

li3 = [1, 2, 3, ['김사과', '오렌지', '반하나', ['🍟','🌭','🥩','🍗']]]
print(li3)
print(li3[2])
print(li3[-1])
print(li3[-1][-2])
print(li3[-1][-1][-2])
print(li3[-1][-1][-1])
````

### 리스트 슬라이싱
- 리스트의 일부분만을 추출할 수 있습니다.
- 슬라이싱은 리스트에서 '여러 개 요소'를 잘라서 새로운 리스트로 만드는 것입니다.
- 결과는 → 잘라낸 "리스트" (대괄호로 감싸진 새로운 리스트)입니다.


```
리스트[start:end:step]

start: 어디서 시작할지 (포함)
end: 어디까지 갈지 (끝 인덱스는 포함하지 않음)
step: 몇 칸씩 이동할지
```

````python
li1 = [10, 20, 30, 40, 50]
print(li1)
print(li1[0:3])

print('--------------------------------------')

li2 = li1
print(li2)
li2[0] = 100
print(li2)
print(li1)

print('--------------------------------------')

li3 = [1, 2, 3, ['김사과', '오렌지', '반하나', ['🍟','🌭','🥩','🍗']]]
print(li3[2:3])
print(li3[3][:2])
````

````python
# 리스트의 항목 추가 / 변경 / 삭제

li4 = [10, 20, 30, 40, 50]
# 슬라이싱을 이용하여 요소를 추가한 경우 리스트에 데이터만 포함
li4[1:2] = ['😁','😂','😎','😍']
print(li4)

print('--------------------------------------')

li4 = [10, 20, 30, 40, 50]
# 인덱싱을 이용하여 요소를 추가한 경우 리스트 안에 리스트를 만들고 포함
li4[1] = ['😁','😂','😎','😍']
print(li4)

print('--------------------------------------')

li4 = [10, 20, 30, 40, 50]
print(li4[1:3])
li4[1:3] = [] # 빈 리스트를 슬라이싱을 통해 저장하면 해당 요소가 삭제됨
print(li4)

print('--------------------------------------')

li4 = [10, 20, 30, 40, 50]
print(li4[1:2])
print(li4[1])
li4[1] = []
print(li4)

print('--------------------------------------')

li4 = [10, 20, 30, 40, 50]
print(li4)
del li4[1]
print(li4)
````

### 리스트의 결합
리스트 간의 + 연산으로 리스트를 결합할 수 있고, * 연산으로 리스트를 반복할 수 있습니다.

````python
li5 = [10, 20, 30]
li6 = [40, 50, 60]
print(li5 + li6)
print(li6 + li5)

print('--------------------------------------')

li5 = [10, 20, 30]
li6 = li5 + [40, 50, 60] # [10, 20, 30] + [40, 50, 60]
print(li6)

print('--------------------------------------')

li6 = [10, 20, 30]
li6 += [40, 50, 60] # li6 = li6 + [40, 50, 60]
print(li6)

print('--------------------------------------')

li6 = [10, 20, 30]
print(li6 * 3)
````

### 함수와 메서드
리스트를 다루기 위한 여러 함수와 메소드들이 있습니다.

````python
# len(): 객체의 길이를 반환하는 파이썬의 기본 내장 함수
li6 = [10, 20, 30]
print(len(li6))
````

````python
# append(): 리스트 요소의 끝에 새로운 요소를 1개만 추가
li6 = [10, 20, 30]
print(li6)

li6.append(100)
print(li6)
# li6.append(200, 300) # TypeError
li6.append([200, 300])
print(li6)
````

````python
# extend(): 리스트 요소의 끝에 새로운 여러 요소를 추가
li6 = [10, 20, 30]
print(li6)

# li6.extend(100) # 무조건 []
li6.extend([100])
print(li6)
li6.extend([200, 300, 400])
print(li6)
````

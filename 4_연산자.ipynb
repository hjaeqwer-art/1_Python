> 연산자는 값을 계산하거나 비교하고, 논리적인 판단을 수행하기 위해 사용하는 기호로, 데이터 간의 연산을 간결하게 표현할 수 있게 해줍니다.

# **1. 산술 연산자**
산술 연산자는 주로 수치 데이터 유형에 대한 기본 산술 연산을 수행하기 위해 사용됩니다.

```
연산자	의미
+	덧셈
-	뺄셈
*	곱셈
/	나눗셈
//	몫을 구하는 나눗셈(Floor Division)
%	나머지
**	거듭제곱
```

````python
print(10 + 3)
print(10 - 3)
print(10 * 3)
print(10 / 3)
print(10 / 2) # 5.0
print(10 // 3) # //는 나눗셈 결과를 내림(Floor) 하여 몫을 구하는 연산자
print(-10 // 3) # -4
print(10 % 3)
print(10 ** 3)
````

**print(10 / 3)이 3.3333333333333335로 출력되는 이유**
> 파이썬의 일반 나눗셈(`/`) 연산이 실수를 반환하는 과정에서, 컴퓨터가 무한소수를 2진수로 처리하다 생긴 '부동소수점 오차' 때문

### 연산자 우선순위
하나의 코드에 여러 연산자가 함께 사용되면 모든 연산을 단순히 왼쪽에서 오른쪽으로 처리하지 않습니다.

수학과 마찬가지로 연산자마다 우선순위가 있습니다.

![연산자 우선순위](./images/연산자%20우선순위.png)

````python
print(10 + 2 * 3)
print((10 + 2) * 3)
````

# **2. 비교 연산자**
비교 연산자는 주로 두 값을 비교하는 데 사용되며, 그 결과는 항상 불리언(True 또는 False) 값입니다.

```

연산자	의미
> 	크다
< 	작다
>=	크거나 같다
<=	작거나 같다
==	같다
!=	같지 않다
```

````python
print(10 > 5)
print(10 < 5)
print(10 >= 5)
print(10 <= 5)
print(10 == 5)
print(10 != 5)

age = 25
print(20 <= age <= 30)
````

# **3. 할당 연산자**
변수에 값을 할당하는데 사용됩니다. 파이썬에서는 기본 할당 연산자 외에도 복합 할당 연산자를 제공하여 코드를 간결하게 작성할 수 있게 도와줍니다.

```
연산자	의미
+=	더한 후 저장
-=	뺀 후 저장
*=	곱한 후 저장
/=	나눈 후 저장
//=	나눈 후 몫만 저장
%=	나눈 후 나머지만 저장
**=	거듭제곱 후 저장
```

````python
num = 10

num += 1 # num = num + 1
print(num)
num -= 1 # num = num - 1 
print(num)
num *= 10 # num = num * 10
print(num)
num /= 10 # / 연산을 수행하므로 결과는 float
print(num)
num %= 2 # 나눈 후 나머지 출력
print(num)
````

# **4. 비트 연산자**
비트 연산자는 정수를 이진 비트로 표현했을 때 사용하는 연산자들입니다. 각 연산자는 정수의 비트 단위로 동작합니다.
```

연산자	의미
&	AND: 전자가 true(1)고, 후자의 값도 true(1)일 때
|	OR: 전자가 true(1)거나, 후자가 true(1)일 때
^	XOR: 전자와 후자가 다를 때 (같으면 0, 다르면 1)
~	NOT: 0은 1로, 1은 0으로 반전 -> 맨 앞이 0이면 양수, 1이면 음수로 인식 => ~n = -(n+1)
<< 	왼쪽 시프트: 전체 비트를 왼쪽으로 지정한 칸수만큼 밀고 빈자리를 0으로 처리
>> 	오른쪽 시프트: 전체 비트를 오른쪽으로 지정한 칸수만큼 밀고 빈자리를 0으로 처리
```

### 이진법 쉽게 구하는 방법
- 1. 주어진 값을 2의 거듭제곱으로 만들어보기
- 2. 주어진 값보다 큰 값이 나오는 2의 거듭제곱부터 0승까지 내림차순으로 정렬
- 3. 거듭제곱 값의 합으로 주어진 값 구해보기
- 4. 값을 구하는 과정에서 사용하지 않은 거듭제곱은 0으로, 사용한 거듭제곱은 1으로 표기 

> x = 50

> 50 / 2^6=64, 2^5=32, 2^4=16, 2^3=8, 2^2=4, 2^1=2, 2^0=1

> -> 50 = 32 + 16 + 2

> -> 50 = 0(64), 1(32), 1(16), 0(8), 0(4), 1(2), 0(1)

> -> 50 = 0110010

````python

a = 60  # 111100
b = 13  # 001101

print(a & b)  # 111100 & 001101 = 001100 => 0 + 0 + 8 + 4 + 0 + 1 = 12
print(a | b)  # 111100 | 001101 = 111101 => 32 + 16 + 8 + 4 + 0 + 1 = 61
print(a ^ b)  # 111100 ^ 001101 = 110001 => 32 + 16 + 0 + 0 + 0 + 1 = 49
print(~a)  # ~a = 000011, ~a = -(a + 1) => ~a = -(60 + 1) = -61
print(a << 2)  # a = 00111100, a << 2 = 11110000 
               # => 128 + 64 + 32 + 16 + 0 + 0 + 0 + 0 = 240
print(a >> 2)  # a = 00111100, a >> 2 = 00001111
               # => 0 + 0 + 0 + 0 + 8 + 4 + 2 + 1 = 15
````

### 음수를 표현하는 방법
컴퓨터는 0과 1로만 데이터를 표현하기 때문에 음수를 직접 표현할 수 없으며, 이를 해결하기 위해 2의 보수(2’s complement) 방식을 사용합니다. 2의 보수는 양수의 이진수를 비트 반전(0↔1)한 후 1을 더하는 방식으로 음수를 표현하는 방법으로, 예를 들어 5(00000101)를 음수로 만들면 비트를 뒤집어 11111010을 만든 뒤 1을 더해 11111011이 되어 -5를 나타냅니다. 이 방식은 0을 하나로 유지할 수 있고, 덧셈만으로 뺄셈을 처리할 수 있어 컴퓨터에서 효율적으로 사용됩니다.

# **5. 논리 연산자**
파이썬에서의 논리 연산자는 주로 불린(Boolean) 값을 조작하기 위해 사용됩니다. 그러나 파이썬의 논리 연산자는 다른 언어와는 다르게, 연산의 결과로 항상 True 또는 False만을 반환하는 것은 아닙니다. 대신 마지막으로 평가된 피연산자의 값을 반환합니다.

```
연산자	의미
and	그리고: 전자가 true고, 후자도 true면 true
or	또는: 둘 중 하나라도 true면 true
not	반대: 값을 반대로 (true면 false, false면 true)
```

````python
a = True
b = False

# AND 연산 (둘 다 True일 때만 True)
print("a and b:", a and b)

# OR 연산 (하나라도 True이면 True)
print("a or b:", a or b)

# NOT 연산 (값을 반대로)
print("not a:", not a)
print("not b:", not b)
````

### 파이썬에서 불린(Boolean)으로 평가되는 주요 값
1. True로 평가되는 주요 값
- 0이 아닌 모든 숫자 (양수, 음수 모두 포함)

- 비어 있지 않은 문자열, 비어 있지 않은 리스트, 비어 있지 않은 튜플, 비어 있지 않은 세트, 비어 있지 않은 딕셔너리

- True

 

2. False로 평가되는 주요 값
- False

- 0

- 빈 문자열

- 빈 리스트 [], 빈 튜플 (), 빈 세트 set(), 빈 딕셔너리 {}

- None

### 단축 평가
단축 평가(Short-Circuit Evaluation)​란 논리 연산에서 결과가 이미 결정되면 뒤의 표현식은 더 이상 실행하지 않는 방식을 말합니다. and 연산자는 앞의 값이 False로 평가되면 전체 결과가 이미 거짓이므로 뒤의 표현식을 확인하지 않고, or 연산자는 앞의 값이 True로 평가되면 전체 결과가 이미 참이므로 뒤의 표현식을 확인하지 않습니다. 이러한 방식은 불필요한 연산을 줄일 뿐만 아니라, 조건에 따라 특정 코드가 실행되지 않도록 제어하는 데에도 유용하게 사용됩니다.

````python
# and
print(True and True) # 전자가 true니까 후자 값(true)을 출력 = True
print(True and False) # 전자가 true니까 후자 값(false)을 출력 = False
print(False and True) # 전자가 false이니까 전자 값(true)을 출력 = False
print(True and 3) # 전자가 true니까 후자 값(3)을 출력 = 3
print(3 and 5) # 전자가 true(0이 아닌 모든 숫자)니까 후자 값(5)을 출력 = 5
print(0 and 5) # 전자가 false(0)이니까 전자 값을 출력 = 0

print('--------------------------------------')

# or
print(True or True) # 전자가 true면 바로 전자 값(true)을 출력 = true
print(True or False) # 전자가 true면 바로 전자 값(true)을 출력 = true
print(False or True) # 전자가 false면 후자 값(true)을 출력 = true
print(False or 3) # 전자가 false면 후자 값(3)을 출력 = 3
print(3 or 5) # 전자가 true(0이 아닌 모든 숫자)니까 바로 전자 값(3)을 출력 = 3
print(0 or 5) # 전자가 false(0)이니까 후자 값(5)을 출력 = 5

print('--------------------------------------')

# not
print(not True) # True를 반전 = False
print(not False) # False를 반전 = True
print(not 0) # False(0)를 반전 = True
print(not 3) # True(0이 아닌 모든 숫자)를 반전 = False
````
